Autograd- automatic differentiation package

```python
# Import pprint, module we use for making our print statements prettier
import pprint
pp = pprint.PrettyPrinter()
```

```python
# requires_grad parameter tells pytorch to store gradients during training we need the gradient and not during the inference

x= torch.tensor([2.], requires_grad=True)

pp.pprint(x.grad)
```

```python
# calculating gradient of y with respect of x
y= x * x * 3 #3x^2

y.backward()
print(x.grad)
pp.pprint(x.grad) # dy/dx = 6x
```

```python
if x.grad is not None:
  x.grad.zero_()
  print(x.grad)

z= x*x*3
z.backward()
pp.pprint(x.grad)
```

```python
block = nn.Sequential(
    nn.Linear(4, 8),     # Layer 1: maps 4 features -> 8
    nn.ReLU(),           # Activation function
    nn.Linear(8, 16),    # Layer 2: maps 8 features -> 16
    nn.ReLU(),
    nn.Linear(16, 2),    # Layer 3: maps 16 features -> 2
    nn.Sigmoid()         # Final output activation
)

input = torch.ones(2, 3, 4)
output = block(input)

pp.pprint(block)
pp.pprint(output)
print("Output shape:", output.shape)
```

```python
class MultilayerPerceptron(nn.Module):

  def __init__(self, input_size, hidden_size):
    super().__init__()

    self.input_size = input_size
    self.hidden_size = hidden_size

    self.model = nn.Sequential(
        nn.Linear(self.input_size, self.hidden_size),
        nn.ReLU(),
        nn.Linear(self.hidden_size, self.hidden_size),
        nn.ReLU(),
        nn.Linear(self.hidden_size, self.input_size),
        nn.Sigmoid()
    )

  def forward(self, x):
      output = self.model(x)
      return output
```

```python
# Or we can also do the above without using sequential as

class MultilayerPerceptron(nn.Module):

  def __init__(self, input_size, hidden_size):
    super().__init__()

    self.input_size = input_size
    self.hidden_size = hidden_size

    self.linear = nn.Linear(self.input_size, self.hidden_size)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(self.hidden_size, self.hidden_size)
    self.relu2 = nn.ReLU()
    self.linear3 = nn.Linear(self.hidden_size, self.input_size)
    self.sigmoid = nn.Sigmoid()

  def forward(self, x):
    linear = self.linear(x)
    relu = self.relu(linear)
    linear2 = self.linear2(relu)
    relu2 = self.relu2(linear2)
    linear3 = self.linear3(relu2)
    output = self.sigmoid(linear3)
    return output
```

```python
# instantiating our class
input = torch.randn(2, 5)
print("Input tensor:", input)

model= MultilayerPerceptron(5, 3)

output = model(input)
pp.pprint(output)

print(output.sum(dim = 1, keepdim=True))
```

```python
import torch.optim as optim
```

```python
y = torch.ones(10, 5)

x= y + torch.randn_like(y)
x
```

```python
model = MultilayerPerceptron(5, 3)

adam = optim.Adam(model.parameters(), lr = 1e-1)

loss_function = nn.MSELoss()

y_pred = model(x)
loss_function(y_pred, y).item()
```

```python
n_epoch= 10

for epoch in range(n_epoch):
  adam.zero_grad()

  y_pred = model(x)

  # print(list(model.parameters()))

  loss = loss_function(y_pred, y)

  print("Epoch %d: train loss: %.4f" % (epoch, loss.item()))

  loss.backward()

  adam.step()
```
